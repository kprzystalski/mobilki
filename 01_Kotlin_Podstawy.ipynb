## 1. Pierwszy program i `println`

`println` wyświetla tekst, a `${...}` osadza wynik wyrażenia w napisie.

```kotlin
run {
    // Wypisanie komunikatu w konsoli
    println("Witaj, Kotlin!")w
    println("2 + 3 = ${2 + 3}") // interpolacja wyrażenia
}
```

## 2. `val`, `var` oraz typy

Stosuj `val` domyślnie, a `var` tylko gdy wartość ma się zmieniać.

```kotlin
run {
    val nazwa: String = "Android" // niemodyfikowalna referencja
    var wersja: Int = 15       // zmienna, którą można zmienić
    val cena: Double = 12.5
    val aktywny: Boolean = true
    wersja += 1
    println("$nazwa $wersja | cena=$cena | aktywny=$aktywny")
}
```

## 3. Operatory i konwersje

Kotlin nie wykonuje automatycznie wszystkich konwersji typów liczbowych.

```kotlin
run {
    val liczbaA = 7
    val liczbaB = 2
    println(liczbaA / liczbaB) // 3: dzielenie całkowitoliczbowe
    println(liczbaA.toDouble() / liczbaB) // 3.5
    println(liczbaA % liczbaB) // reszta z dzielenia
    println(liczbaA > liczbaB && liczbaB > 0)
}
```

## 4. Instrukcje `if` i `when`

`if` i `when` są wyrażeniami: mogą zwracać wartości.

```kotlin
run {
    val punkty = 81
    val ocena = if (punkty >= 90) "bardzo dobry"
        else if (punkty >= 75) "dobry"
        else "do poprawy"

    val kategoria = when (punkty) {
        in 0..49 -> "niezaliczone"
        in 50..74 -> "zaliczone"
        in 75..100 -> "wysoki wynik"
        else -> "wynik niepoprawny"
    }
    println("$ocena / $kategoria")
}
```

## 5. Pętle i zakresy

Poznaj `..`, `until`, `downTo` i `step`.

```kotlin
run {
    for (i in 1..5) print("$i ")
    println()
    for (i in 10 downTo 2 step 2) print("$i ")
    println()
    var suma = 0
    var i = 1
    while (i <= 5) {
        suma += i
        i++
    }
    println("Suma: $suma")
}
```

## 6. Funkcje i parametry domyślne

W Kotlinie można używać krótkich funkcji z `=` i argumentów nazwanych.

```kotlin
run {
    fun poleProstokata(a: Double, b: Double): Double = a * b
    fun powitanie(imie: String, jezyk: String = "pl"): String =
        if (jezyk == "pl") "Cześć, $imie!" else "Hello, $imie!"

    println(poleProstokata(3.0, 4.0))
    println(powitanie("Ola"))
    println(powitanie(imie = "Tom", jezyk = "en")) // argumenty nazwane
}
```

## 7. Null safety i operator Elvis

`String` nie dopuszcza `null`, a `String?` dopuszcza. Unikaj operatora `!!` bez potrzeby.

```kotlin
run {
    val nick: String? = null // dopuszczamy null
    println(nick?.length) // bezpieczny dostęp
    println(nick?.length ?: 0) // wartość domyślna

    val email: String? = "student@uczelnia.pl"
    if (email != null) println(email.uppercase())
}
```

## 8. Łańcuchy znaków

Łańcuchy są niemutowalne; metody zwracają nowe wartości.

```kotlin
run {
    val zdanie = "  Kotlin jest przydatny w Androidzie  "
    println(zdanie.trim().uppercase())
    println("Długość: ${zdanie.length}")
    println("Zawiera Android: ${"Android" in zdanie}")
    val slowa = zdanie.trim().split(" ")
    println(slowa)
}
```

## 9. Listy: niemutowalne i mutowalne

`listOf` udostępnia interfejs tylko do odczytu; `mutableListOf` pozwala modyfikować zawartość.

```kotlin
run {
    val miasta = listOf("Kraków", "Warszawa", "Gdańsk")
    val zadania = mutableListOf("Projekt", "Testy")
    zadania.add("Publikacja")
    println(miasta.first())
    println(zadania)
    println(zadania.size)
}
```

## 10. Mapy i zbiory

Odczyt brakującego klucza w mapie zwraca `null`.

```kotlin
run {
    val oceny = mapOf("Anna" to 5, "Jan" to 4)
    val unikalne = setOf("Kotlin", "Swift", "Kotlin")
    println(oceny["Anna"])
    println(oceny["Brak"] ?: "brak oceny")
    println(unikalne) // zbiory nie przechowują duplikatów
}
```

## 11. Kolekcje funkcyjne `filter`, `map`, `sum`

Operacje na kolekcjach pozwalają budować czytelne potoki przetwarzania danych.

```kotlin
run {
    val wyniki = listOf(45, 81, 67, 93, 50)
    val zaliczone = wyniki.filter { it >= 50 }
    val podwyzszone = zaliczone.map { it + 5 }
    println(zaliczone)
    println(podwyzszone)
    println("Suma: ${zaliczone.sum()}")
}
```

## 12. Lambdy i funkcje wyższego rzędu

Lambda to funkcja anonimowa, którą można przekazać jako argument.

```kotlin
run {
    fun policz(a: Int, b: Int, operacja: (Int, Int) -> Int): Int =
        operacja(a, b)

    val mnozenie: (Int, Int) -> Int = { x, y -> x * y }
    println(policz(4, 3, mnozenie))
    println(policz(4, 3) { x, y -> x + y })
}
```

## 13. Klasy i obiekty

`private set` blokuje modyfikowanie właściwości spoza klasy.

```kotlin
run {
    class Licznik(start: Int = 0) {
        var wartosc: Int = start
            private set

        fun zwieksz() { wartosc++ }
    }
    val licznik = Licznik(10)
    licznik.zwieksz()
    println(licznik.wartosc)
}
```

## 14. `data class` oraz `copy`

`data class` automatycznie dostarcza m.in. `toString`, `equals` i `copy`.

```kotlin
run {
    data class Student(val imie: String, val punkty: Int)
    val student = Student("Marta", 78)
    val poprawiony = student.copy(punkty = 88)
    println(student)
    println(poprawiony)
    println(student == poprawiony)
}
```

## 15. Dziedziczenie i interfejsy

Interfejs definiuje kontrakt, a klasa dostarcza implementację.

```kotlin
interface Powiadamialny {
    fun powiadom(): String
}

class PowiadomienieEmail(val adres: String) : Powiadamialny {
    override fun powiadom(): String =
        "Wysyłam e-mail na $adres"
}

fun main() {
    val p: Powiadamialny =
        PowiadomienieEmail("student@example.com")

    println(p.powiadom())
}

main()
```

## 16. `enum class` i `when`

Dzięki `enum` reprezentujemy zamknięty zbiór wartości.

```kotlin
enum class Priorytet {
    NISKI, SREDNI, WYSOKI
}

fun opis(p: Priorytet): String = when (p) {
    Priorytet.NISKI -> "może poczekać"
    Priorytet.SREDNI -> "zaplanować"
    Priorytet.WYSOKI -> "wykonać teraz"
}

fun main() {
    println(opis(Priorytet.WYSOKI))
}

main()
```

## 17. Obsługa wyjątków

`try` również może zwrócić wartość. Zwykle preferuj funkcje takie jak `toIntOrNull`.

```kotlin
run {
    fun bezpieczneDzielenie(a: String, b: String): Int? = try {
        a.toInt() / b.toInt()
    } catch (e: NumberFormatException) {
        null // niepoprawna liczba
    } catch (e: ArithmeticException) {
        null // dzielenie przez zero
    }
    println(bezpieczneDzielenie("20", "4"))
    println(bezpieczneDzielenie("20", "0"))
    println("abc".toIntOrNull()) // alternatywa bez wyjątku
}
```

## 18. Rozszerzenia (extension functions)

Funkcje rozszerzające dodają wygodną składnię bez modyfikacji kodu klasy.

```kotlin
run {
    fun String.pierwszaWielka(): String =
        replaceFirstChar { it.uppercase() }
    println("kotlin".pierwszaWielka())
}
```

## 19. Wprowadzenie do modelu danych aplikacji

Takie modele danych mogą później służyć jako podstawa list w Jetpack Compose.

```kotlin
run {
    data class Zadanie(val id: Int, val tytul: String, val ukonczone: Boolean)
    val zadania = listOf(
        Zadanie(1, "Zaprojektuj ekran", true),
        Zadanie(2, "Dodaj przycisk", false),
        Zadanie(3, "Napisz test", false)
    )
    val pozostale = zadania.filterNot { it.ukonczone }
    println("Do wykonania: ${pozostale.size}")
    pozostale.forEach { println("• ${it.tytul}") }
}
```

## 20. Proste sprawdzanie poprawności kodu

`check` sygnalizuje błąd, jeśli warunek jest fałszywy. To wstęp do testów jednostkowych.

```kotlin
run {
    fun czyPelnoletni(wiek: Int): Boolean = wiek >= 18
    check(czyPelnoletni(20))
    check(!czyPelnoletni(16))
    println("Wszystkie asercje przeszły")
}
```
